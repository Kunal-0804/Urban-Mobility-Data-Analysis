##  Date and Time Validation

The date and time columns are validated after cleaning to ensure that they remain in appropriate datetime formats for time-based analysis.

```python
# Definition: This removes leading and trailing whitespace only from text columns.

text_columns = [
    "Booking ID",
    "Booking Status",
    "Customer ID",
    "Vehicle Type",
    "Pickup Location",
    "Drop Location",
    "Reason for cancelling by Customer",
    "Driver Cancellation Reason",
    "Incomplete Rides Reason",
    "Payment Method"
]

for col in text_columns:
    cleaned_df[col] = cleaned_df[col].str.strip()

print("Text columns standardized successfully.")
```

```python
# Definition: This combines the booking date and booking time into a single datetime value.

cleaned_df["Booking Datetime"] = pd.to_datetime(
    cleaned_df["Date"].astype(str) + " " + cleaned_df["Time"].astype(str)
)

print("Booking Datetime created successfully.")
print("Booking Datetime dtype:", cleaned_df["Booking Datetime"].dtype)
```

## Create Booking Datetime

The date and time columns are combined into a single datetime column to support time-based analysis and visualization.

```python
# Summary:
# This code uses NumPy to group different booking statuses
# into standardized ride outcome categories for easier analysis.

# Definition: This creates a standardized ride outcome category using NumPy conditional logic.

conditions = [
    cleaned_df["Booking Status"] == "Completed",
    cleaned_df["Booking Status"] == "Incomplete",
    cleaned_df["Booking Status"] == "Cancelled by Customer",
    cleaned_df["Booking Status"] == "Cancelled by Driver",
    cleaned_df["Booking Status"] == "No Driver Found"
]

choices = [
    "Completed",
    "Incomplete",
    "Customer Cancelled",
    "Driver Cancelled",
    "No Driver Found"
]

cleaned_df["Ride Outcome Category"] = np.select(
    conditions,
    choices,
    default="Other"
)

cleaned_df["Ride Outcome Category"].value_counts()
#np.select() multiple conditions ko check karta hai.
```

##  Create Ride Outcome Category

The booking status is grouped into standardized ride outcome categories using NumPy conditional logic for easier analysis and visualization.

```python
# Definition: This compares the original booking status with the new ride outcome category.

pd.crosstab(
    cleaned_df["Booking Status"],
    cleaned_df["Ride Outcome Category"]
)
```

```python
# Definition: This extracts useful time-based features from the booking datetime.

cleaned_df["Booking Month"] = cleaned_df["Booking Datetime"].dt.month
cleaned_df["Booking Hour"] = cleaned_df["Booking Datetime"].dt.hour
cleaned_df["Booking Day"] = cleaned_df["Booking Datetime"].dt.day_name()

print("Time-based columns created successfully.")

cleaned_df[
    ["Booking Datetime", "Booking Month", "Booking Hour", "Booking Day"]
].head()
```
